```python
# =======================================================
# 2. データ取得と統合
# =======================================================
# Rossmann Store Sales(kaggle)のデータ読込み
train = pd.read_csv("/content/train.csv")
test = pd.read_csv("/content/test.csv")
store = pd.read_csv("/content/store.csv")

# 店舗情報を結合
df = train.merge(store, on="Store", how="left")

# サンプルとして特定の1店舗（Store 7）に絞り込み
# 理由：店舗ごとの傾向の違いを排除し、モデルの基礎ロジックを検証するため
df = df[df["Store"] == 7]
df = df.sort_values("Date").reset_index(drop=True)
```

```python
# 予測対象は営業日のみとするため、閉店日を除去
df = df[df["Sales"] > 0]
```

```python
# 時系列トレンド（月・曜日別売上）
# 月や曜日ごとの季節性・週次トレンド
# 特定の曜日や月に売上が偏っていないか

# 日付をdatetimeに変換
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
df['DayOfWeek'] = df['Date'].dt.dayofweek

# 月別平均売上
monthly_sales = df.groupby('Month')['Sales'].mean()
monthly_sales.plot(kind='bar', figsize=(8,5), title='月別平均売上')
plt.show()

# 曜日別平均売上
dow_sales = df.groupby('DayOfWeek')['Sales'].mean()
dow_sales.plot(kind='bar', figsize=(8,5), title='曜日別平均売上')
plt.show()
```

```python
# 日付カラム変換
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")
```

```python
# dfの欠損値確認
print(df.isna().sum())
```

```python
# =======================================================
# 4 特徴量作成
# =======================================================

df["month"] = df["Date"].dt.month
df["day_of_week"] = df["Date"].dt.dayofweek

# ラグ特徴量
df["lag_14"] = df["Sales"].shift(14)
df["lag_28"] = df["Sales"].shift(28)

# 過去14日間、30日間の売上平均
df["rolling_mean_14"] = df["Sales"].shift(14).rolling(window=14).mean()
df["rolling_mean_30"] = df["Sales"].shift(30).rolling(window=30).mean()

# 過去14日間の売上標準偏差
df["rolling_std_14"] = df["Sales"].shift(14).rolling(14).std()


# 欠損行を削除（ラグ・移動統計量の初期期間がNaNになるため）
df = df.dropna(subset=[  "month","day_of_week","Promo",
    "SchoolHoliday","lag_14", "lag_28", "rolling_mean_14",
    "rolling_mean_30","rolling_std_14",
]).reset_index(drop=True)

# 確認用
print(f"特徴量作成後の残りデータ件数: {len(df)} 件")
```

```python
# データ確認
df.head()
```

```python
# =======================================================
# 5 学習データ
# =======================================================

features = [
    "month",
    "day_of_week",
    "Promo",
    "SchoolHoliday", # 学校休日日
    "lag_14",
    "lag_28",
    "rolling_mean_14",
    "rolling_mean_30",
    "rolling_std_14",
]

X = df[features]
y = df["Sales"]
```

```python
# =======================================================
# 6 Expanding Window CV
# =======================================================

horizon = 14
n_folds = 3
val_size = horizon  # 直近14日分を early_stopping 用の検証データにする

rmse_list = []
mae_list = []
rel_mae_list = []
rmse_pct_list = []

last_y_test = None
last_pred = None

for i in range(n_folds):
    # Foldごとに horizon 日ずつテストセットをスライド
    train_end = len(df) - horizon * (n_folds - i)

    # train_endが十分かチェック
    if train_end <= 0:
        print(f"Fold{i+1} skipped: train_end={train_end}")
        continue

    X_train = X.iloc[:train_end]
    y_train = y.iloc[:train_end]

    X_test = X.iloc[train_end:train_end + horizon]
    y_test = y.iloc[train_end:train_end + horizon]

    # train<val_sizeまたはtestが空ならスキップ
    if len(X_train) <= val_size or len(X_test) == 0:
        print(f"Fold{i+1} skipped: insufficient train or test data")
        continue

    # 学習データを「実学習用(tr)」と「Early Stopping検証用(val)」に時系列順で分割
    X_tr, y_tr = X_train.iloc[:-val_size], y_train.iloc[:-val_size]
    X_val, y_val = X_train.iloc[-val_size:], y_train.iloc[-val_size:]

    # ----------------------
    # LightGBM モデル
    # ----------------------
    model = lgb.LGBMRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=5,
        importance_type="gain",
        verbosity=-1,
        random_state=42
    )

    model.fit(
        X_tr,
        y_tr,
        eval_set=[(X_val, y_val)],
        eval_metric="rmse",
        callbacks=[lgb.early_stopping(50, verbose=False)]
    )

    # ----------------------
    # 未見のテストデータで評価
    # ----------------------
    pred = model.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test, pred))
    mae = mean_absolute_error(y_test, pred)

    # テスト期間における lag_14 ナイーブ予測とのMAE比
    naive_pred = X_test["lag_14"].values
    baseline_mae = mean_absolute_error(y_test, naive_pred)
    rel_mae = mae / baseline_mae
    improvement = (1 - rel_mae) * 100

    rmse_pct = rmse / y_test.mean() * 100

    rmse_list.append(rmse)
    mae_list.append(mae)
    rel_mae_list.append(rel_mae)
    rmse_pct_list.append(rmse_pct)

    # 各Fold結果を表示
    print(f"Fold{i+1} RMSE(%)         : {rmse_pct:.2f}%")
    print(f"Fold{i+1} MAE             : {mae:.2f}")
    print(f"Fold{i+1} 対Lag14 MAE比   : {rel_mae:.3f}")
    print(f"Fold{i+1} 対Lag14 改善率  : {improvement:.1f}% 改善")
    print()

    # 最後のfoldだけ保存
    if i == n_folds - 1:
        last_y_test = y_test
        last_pred = pred

# =======================================================
# 7 評価表示
# =======================================================
print()
print("CV平均結果（Store 7 / 14日×3フォールド）")

mean_rmse_pct = np.mean(rmse_pct_list)
mean_mae = np.mean(mae_list)
mean_rel_mae = np.mean(rel_mae_list)
mean_improvement = (1 - mean_rel_mae) * 100

print(f"CV平均 RMSE(%)       : {mean_rmse_pct:.2f}%")
print(f"CV平均 MAE           : {mean_mae:.2f}")
print(f"CV平均 対Lag14 MAE比 : {mean_rel_mae:.3f}")
# CV平均 対Lag14 MAE比 < 1 なら naive予測（lag_14）より良い
print(f"CV平均 対Lag14 改善率: {mean_improvement:.1f}% 改善")

# =======================================================
# 8 予測グラフ
# =======================================================
if last_y_test is not None and last_pred is not None:
    plt.figure(figsize=(12, 6))
    x = np.arange(1, len(last_pred) + 1)

    plt.plot(x, last_y_test.to_numpy(), label="実績")
    plt.plot(x, last_pred, label="予測")

    plt.title("LightGBMによる需要予測（最後のFold / Store 7）")
    plt.xlabel("予測日数")
    plt.ylabel("売上")
    plt.legend()
    plt.show()
```

```python
# =======================================================
# 9 特徴量重要度
# =======================================================

importance = pd.DataFrame({
    "feature": features,
    "importance": model.feature_importances_
}).sort_values("importance", ascending=False)

plt.figure(figsize=(8,5))

sns.barplot(data=importance, x="importance", y="feature")

plt.title("特徴量重要度")

plt.show()
```

### Feature Importance（特徴量重要度）

LightGBM モデル（最終Fold・`importance_type="gain"`）で予測に寄与した特徴量を可視化しました（対象：Rossmann Store 7）。

* **プロモーションの影響**：`Promo` などのプロモーション情報が予測に強く影響しています。
* **カレンダー情報**：曜日（`day_of_week`）・月（`month`）・学校休日（`SchoolHoliday`）などのカレンダー情報も一定の重要度を持ち、週次パターンや季節性の捉え方に貢献しています。
* **過去の売上データ**：`rolling_mean_14`、`rolling_mean_30`、`lag_14`、`lag_28`、`rolling_std_14` などのラグ・移動統計量も予測に寄与していますが、今回の単一店舗（Store 7）設定ではプロモーションやカレンダー情報に比べて相対的な重要度はやや低めでした。

#### この分析から得られる示唆
* プロモーション実施有無（`Promo`）が売上変動の大きなドライバーになっている
* 曜日・月・休日といったカレンダー要因を組み合わせることで、基礎的な需要リズムを捉えられる
* 過去実績のラグ・移動平均特徴量も補助的に効いている
